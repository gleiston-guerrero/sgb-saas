## 3. Fórmula de Brooke (1996) — protocolo, sin corrida

Ítems **impares** (Q1, Q3, Q5, Q7, Q9): contribución = `valor - 1`. Ítems **pares** (Q2, Q4, Q6, Q8, Q10): contribución = `5 - valor`. Score = `(suma) × 2.5` → rango 0–100. La función queda como implementación de referencia; **no se invoca**, porque no existen respuestas.

```python
def calcular_score_sus(q1, q2, q3, q4, q5, q6, q7, q8, q9, q10):
    """Score SUS (Brooke 1996) para Q1..Q10 en escala Likert 1-5.

    Referencia del protocolo. Sin invocaciones en este entregable (N=0).
    """
    vals = (q1, q2, q3, q4, q5, q6, q7, q8, q9, q10)
    assert all(isinstance(v, int) and 1 <= v <= 5 for v in vals)
    suma = sum(v - 1 if i % 2 == 0 else 5 - v for i, v in enumerate(vals))
    return suma * 2.5


assert callable(calcular_score_sus)
print('Funcion de referencia definida:', calcular_score_sus.__name__)
print('Invocaciones sobre respuestas: 0 (N=0, sin datos).')
```
